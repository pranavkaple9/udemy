```python
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt
```

```python

X, y = make_blobs(n_samples=40, centers=2, random_state=6)
```

```python
X.shape, y.shape
```

```python
# this gives all the X values for feature 0 where y is 1
X[y][:,0]
```

```python
y_green = y == 1
plt.figure(figsize=(12,8))

# We plot for y = 0 in red here x axis is feature 0 and y axis is feature 1
plt.scatter(X[~y_green][:, 0], X[~y_green][:, 1], color='red')

# We plot for y = 1 in green here x axis is feature 0 and y axis is feature 1
plt.scatter(X[y_green][:, 0], X[y_green][:, 1], color='green')

# this is useful in case of SVM where we need to separate the two classes to see which kernel works best
```
