## Handling Imbalanced Dataset

1. Up Sampling
2. Down Sampling

```python
import numpy as np
import pandas as pd

# Set the random seed for reproducibility
np.random.seed(123) # it means that every time you run the code, you'll get the same output. np.random.rand(3) will always produce the same three random numbers.

# Create a dataframe with two classes
n_samples = 1000
class_0_ratio = 0.9
n_class_0 = int(n_samples * class_0_ratio)
n_class_1 = n_samples - n_class_0
```

```python
n_class_0,n_class_1
```

```python
## CREATE MY DATAFRAME WITH IMBALANCED DATASET
class_0 = pd.DataFrame({
    'feature_1': np.random.normal(loc=0, scale=1, size=n_class_0), #normal distribution with mean 0 and standard deviation 1
    'feature_2': np.random.normal(loc=0, scale=1, size=n_class_0),
    'target': [0] * n_class_0
})

class_1 = pd.DataFrame({
    'feature_1': np.random.normal(loc=2, scale=1, size=n_class_1), #normal distribution with mean 2 and standard deviation 1
    'feature_2': np.random.normal(loc=2, scale=1, size=n_class_1),
    'target': [1] * n_class_1
})
class_0.head()
```

```python
df=pd.concat([class_0,class_1]).reset_index(drop=True) # to concat both must have same columns and reset index to avoid duplicate index
```

```python
df.tail()
```

```python
df['target'].value_counts()
```

```python
## upsampling means to increase the number of samples in the minority class by duplicating existing samples or generating new ones.
df_minority=df[df['target']==1]
df_majority=df[df['target']==0]
```

```python
!pip install scikit-learn
```

```python
from sklearn.utils import resample
# n_samples=len(df_majority) means we want to upsample minority class to have same number of samples as majority class
# df_minority_upsampled will have 900 rows, all from the original df_minority, but with duplicates due to sampling with replacement
# If replace=False, the resample function will not allow duplicate rows when sampling. so if the original df_minority < n_samples, it will raise an error.
# if rows in df_minority == n_samples, then replace can be either True or False, it will give same result.
df_minority_upsampled=resample(df_minority,replace=True, #Sample With replacement
         n_samples=len(df_majority),
         random_state=42
        )
```

```python
df_minority_upsampled.shape
```

```python
df_minority_upsampled.head()
```

```python
df_upsampled=pd.concat([df_majority,df_minority_upsampled]) # axis=0 by default, so it will concat row-wise
```

```python
df_upsampled.shape
```

```python
df_upsampled['target'].value_counts()
```

## Down Sampling

```python
import pandas as pd

# Set the random seed for reproducibility
np.random.seed(123)

# Create a dataframe with two classes
n_samples = 1000
class_0_ratio = 0.9
n_class_0 = int(n_samples * class_0_ratio)
n_class_1 = n_samples - n_class_0

class_0 = pd.DataFrame({
    'feature_1': np.random.normal(loc=0, scale=1, size=n_class_0),
    'feature_2': np.random.normal(loc=0, scale=1, size=n_class_0),
    'target': [0] * n_class_0
})

class_1 = pd.DataFrame({
    'feature_1': np.random.normal(loc=2, scale=1, size=n_class_1),
    'feature_2': np.random.normal(loc=2, scale=1, size=n_class_1),
    'target': [1] * n_class_1
})

df = pd.concat([class_0, class_1]).reset_index(drop=True)

# Check the class distribution
print(df['target'].value_counts())
```

```python
## downsampling
df_minority=df[df['target']==1]
df_majority=df[df['target']==0]
```

```python
from sklearn.utils import resample
df_majority_upsampled=resample(df_minority,replace=True, #Sample With replacement
         n_samples=len(df_majority),
         random_state=42
        )
```
